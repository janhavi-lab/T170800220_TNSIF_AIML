Common Operations in Lists:

Operation	Description
append()	Add an element at the end
extend()	Add multiple elements from another iterable
insert()	Insert an element at a specified position
remove()	Remove the first occurrence of a specified value
pop()	Remove and return an element at a given index
clear()	Remove all elements from the list
index()	Return the first index of a specified value
count()	Count the number of occurrences of a value
reverse()	Reverse the list in-place
sort()	Sort the list in ascending order (by default)
copy()	Return a shallow copy of the list
len()	Return the number of elements in the list
sum()	Return the sum of all numeric elements
max() / min()	Return the largest or smallest element

```python
import builtins

nums = []

for i in range(5):
  val = float(input(f"Enter number {i+1}: "))
  nums.append(val)

# Calculate the sum and average of the numbers
total_sum = builtins.sum(nums)
average = total_sum / len(nums)

print(f"The entered numbers are: {nums}")
print(f"The average of the numbers is: {average}")
print(f"The sum of the numbers is: {total_sum}")
```

```python
#Count the number of digits in a number.
num = 221325;
count=0;
while(num>0):
  count = count + 1;
  num = num//10;
print(count);
```

```python
#Reverse a number using a loop.
num = 25;
reverse =0;
while num>1:
  rem = num%10;
  reverse = reverse*10 + rem;
  num = num//10;
print(reverse)
```

```python
#Check whether a number is prime.
import math
num = 37;
if num<=1:
  print("Non Prime")

isPrime=True;

for i in range(2,int(math.sqrt(num) + 1)):
  if num%i==0:
    isPrime = False;
    break;
  else:
    isPrime = True;

if isPrime:
  print("Prime")
else:
  print("Non Prime")
```

```python
#Write a function to check if a number is palindrome.
num = 12321;
rev = 0;
temp = num;
while num>0:
  rem = num%10;
  rev = rev*10 + rem;
  num = num//10;
if temp == rev:
  print("Palindrome")
else:
  print("Non Palindrome")
```

```python
#Write a function to calculate sum of digits of a number.
num = 123;
sum=0;
while num>0:
  rem = num%10;
  sum = sum + rem;
  num = num//10;
print(sum)
```
